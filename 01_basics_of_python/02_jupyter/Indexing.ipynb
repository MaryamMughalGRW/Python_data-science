# INDEXING  IN PYTHON AND MATRICES

```python
# pyhthon mn left to right 0 sy index starting hota hai r sath mn neg mn apny dekhna hai to right sy count karna hota hai matlab last elemnet ka index -1 hota hai aur second last ka -2 hota hai aur is tarah se count karte hain
tuple1 = (1, 2, 3, 4, 5)
print(tuple1[0])  # Output: 1
print(tuple1[-1])  # Output: 5
print(tuple1[2])  # Output: 3
print(tuple1[-2])  # Output: 4
```

```python
# matrics mn phli row ka index 0 hota hai aur phli column ka index 0 hota hai aur is tarah se count karte hain
matrix=[[1,2,3],
        [5,8,9],
        [4,5,6]
        ]
        # but isko 3*3 ka hi matrix kehty hain
print(matrix[0][0])  # Output: 1
print(matrix[0][2])  # Output: 3
print(matrix[2][2])  # Output: 6
```

```python
name="Maryam Mughal"
print(len(name)) #13 lenght ayi gi q ky space bi count hoti haii but index 0 sy start hota hai to last index 12 hoga
print(name[8])
print(name[-1]) #last charachter print hoga
print(name[0])
print(name[12])
print(name[0:12]) #0 sy 11 tk print hoga， 12 index include ni hoga
print(name[0:13]) #0 sy 12 tk print hoga r 13 index include ni hoga 
print(name[7])
print(name[6])#space hai yahan
print(name[-2])
 # print(name[13]) #error ay gi q ky 13 index exist nh krta
```

# SLICING 

```python
roll_number=[1,2,3,4,5,6,7,8,9]
print(roll_number[0:5]) #0 sy 4 tk print hoga, 5 index include nh hoga
print(roll_number[0:-2]) #0 sy -2 tk print hoga, -2 index include nh hoga
print(roll_number[-6:-1]) #-6 sy -1 tk print hoga, -1 index include nh hoga ,retrive esy hi kiya jata hai is mn phly -6 phr -1 index likha jata hai;-1:-6 sy errr aye ga yh smj lo jesy number line sy data uthana haii
print(roll_number[::3])# 3 values ky bad slice kary ga r jo aye ga next usy print kary ga 1st wala zero index include hoga
```

# MUTATABLE ELEMENTS

```python
room_no=[101,102,103,104,105,106,107,108,109]
room_no[3]=201 #3 index replace hoga
room_no[4:-1]=[201,202,203,204] #4 sy -1 tk replace hoga, -1 index include nh hoga
print(room_no)
```

# Using Range Function in list

```python
total_students=list(range(1,10))
print(total_students)
class_no=list(range(1,100,10))
print(class_no)
house_no=list(range(0,100,10))
print(house_no)
total_unis=list(range(1,100)) #100 include ni hoga
print(total_unis)
total_hospials=list(range(0,100))
print(total_hospials) 
total_schools=list(range(1,101,))
print(total_schools) #101 include nh hoga
```
