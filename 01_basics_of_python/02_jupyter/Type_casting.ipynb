# Type CASTING


```python
a_int=25
a_float=float(a_int)
print(a_float)
```

```python
#float to int
a_float=25.7
a_int=int(a_float)
print(a_int)
```

```python
# float or int to string
a_int=2500
a_float=2500.0

a_str=str(a_int)  
 #jupiter variable explorer will show the type of a_str and a_str2 as string
a_str2=str(a_float)  
print(a_str,a_str2)
```

```python
# string to int or float
a_str="25"
# This will raise a ValueError because "maryam" cannot be converted to an integer or float
a_int=int(a_str)
a_float=float(a_str)
print(a_int,a_float)
```

```python
#list to tuple and vice versa
c_list=[1,2,3,4,5]
c_tuple=tuple(c_list)
print(c_list)
print(c_tuple)
```

```python
# list to set and set to list
c_set = set(c_list)
c_list_again = list(c_set)

print(c_set)
print(c_list_again)
```

```python
b= 10
b_bool=bool(b)
print(b_bool)
#any non-zero number is considered True, while zero is considered False. Therefore, when you convert the integer 10 to a boolean using bool(10), it evaluates to True.
empty_list=[]
empty_list_bool=bool(empty_list)        
print(empty_list_bool)
```

```python
# Note: eval() executes input as code!
# Numbers (45) become int, quotes ("kilo") become string, but plain text (kilo) looks for a variable name!
a = eval(input("Enter something: "))
print(f"The value you entered is: {a} and it's type is {type(a)}")
```

```python
m=input("Enter a number: ")
#input by default takes input as string, so we need to convert it to int or float if we want to perform mathematical operations on it.
print(f"The number you entered is: {m} and it's type is {type(m)}")
```


# taking input from user and converting it to int or float

```python
user_input = input("Enter your weight/age: ")

try:
    # 1. String ko Float mein convert karne ki koshish
    val = float(user_input)
    
    # 2. Agar convert ho gaya to Math Operation
    result = val * 2
    print(f"Success! Calculated result is: {result}")

except ValueError:
    # 3. Agar text dala gaya to ValueError yahan catch hoga
    print("Invalid Input! Aap ne text daal diya hai, sirf numeric number daalein.")
```
