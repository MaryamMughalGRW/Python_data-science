# IF-ELSE STATEMENT

```python
a=input("enter your marks:")
if a>="85":
    print("GPA:4")
elif a>="80" and a<"86":
    print("GPA:3.7")
elif a>="75" and a<"80":
    print("GPA:3.3")
elif a>="70" and a<"75":
    print("GPA:3.0")
elif a>="65" and a<"70":
    print("GPA:2.7")
elif a>="60" and a<"65":
    print("GPA:2.3")
else:
    print("GPA:2.0")
```
