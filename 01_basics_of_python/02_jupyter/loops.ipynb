# FOR AND WHILE AND FOR-ELSE LOOP

```python
for i in range(1,11):
    multiple=7*i
    print(f'7 * {i} = {multiple}')
```

```python
numbers={1,2,3,4,5,6}
for number in numbers:
    print("*"*number)
    
```

```python
names=['hira','','sarah','maria']
for name in names:
    if name=='':
        print("no name found:",name)
        break
else:
    print("all names are valid")    
```

```python
files=['file1.txt','file2.pdf','file3.txt']
for file in files:
    if not file.endswith('.txt'):
        print("not a text file:",file)
        break
else:
    print("all files are text files")
```

```python
# crossing data
colors={'red','blue','green'}
sizes={'S','M','L'}
for color in colors:
    for size in sizes:
        print(f'{color} size:{size}')
```
