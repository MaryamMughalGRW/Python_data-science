# Types of variables

```python
# Common Python data types

integer_value = 42
#immutable whole number
float_value = 3.14
# immutable floating-point number
complex_value = 2 + 3j
# immutable complex number
string_value = "Hello, Python!"
# immutable string
boolean_value = True
# immutable boolean
list_value = [1, 2, 3]
#  mutable ordered collection of items
tuple_value = (1, 2, 3)
# immutable ordered collection of items
set_value = {1, 2, 3}
# unordered mutable collection of unique items
dictionary_value = {"name": "Ali", "age": 25}
# collection of mutable key-value pairs
none_value = None
# represents the absence of a value

examples = {
    "integer": integer_value,
    "float": float_value,
    "complex": complex_value,
    "string": string_value,
    "boolean": boolean_value,
    "list": list_value,
    "tuple": tuple_value,
    "set": set_value,
    "dictionary": dictionary_value,
    "None": none_value,
}

for name, value in examples.items():
    print(f"{name}: {value!r} (type: {type(value).__name__})")
```

# WHAT DOES IMMUTABILITY MEANS:

```python
# String Immutable ہے
name = "Hira"

# یہ لائن ایرر (TypeError) دے گی:
try:
    name[0] = "M" 
except TypeError as e:
    print(f"Error: {e}")
    print("ثابت ہوا کہ سٹرنگ کی اندرونی ویلیو تبدیل نہیں ہو سکتی!")
```

```python
# Reassignment میں نیا ابجیکٹ بنتا ہے
name = "Hira"
print("پہلا میموری ایڈریس:", id(name))

name = "Hira Khan"    
# اب نیا سٹرنگ بنا ہے، پرانی جگہ پر تبدیلی نہیں ہوئی
print("نیا میموری ایڈریس:", id(name))
print("ایڈریس بدل گیا، یعنی پرانی جگہ پر تبدیلی نہیں ہوئی بلکہ نیا سٹرنگ بنا!") 
```

# What does mutable means:

```python
# List Mutable ہے
my_list = ["H", "i", "r", "a"]
print("تبدیلی سے پہلے ایڈریس:", id(my_list))

# پہلا ایلیمنٹ بدلیں
my_list[0] = "M"
print("تبدیل شدہ لسٹ:", my_list)
print("تبدیلی کے بعد ایڈریس:", id(my_list))
print("ایڈریس وہی رہا، یعنی لسٹ اسی جگہ (in-place) تبدیل ہوئی ہے!")
```

# Ordered vs unordered

```python
# List (Ordered) - انڈیکس کے ذریعے حاصل ہو سکتی ہے
my_list = ["Apple", "Banana", "Cherry"]
print("لیسٹ کا پہلا ایلیمنٹ (Index 0):", my_list[0])  # Apple

# Set (Unordered) - انڈیکس نہیں ہوتا
my_set = {"Apple", "Banana", "Cherry"}

try:
    print(my_set[0])
except TypeError as e:
    print(f"\nSet Error: {e}")
    print("ثابت ہوا کہ Set میں انڈیکس کا کوئی وجود نہیں، یہ Unordered ہے!")
```
